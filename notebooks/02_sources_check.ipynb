```python
import pandas as pd
from churnguard.data import load_clean_data

local = load_clean_data("local").sort_values("customer_id").reset_index(drop=True)
snow = load_clean_data("snowflake").sort_values("customer_id").reset_index(drop=True)

print(local.shape, snow.shape)
print("Null total_charges:", local["total_charges"].isna().sum(), snow["total_charges"].isna().sum())

pd.testing.assert_frame_equal(local, snow, check_dtype=False)
print("✅ Local and Snowflake data are identical")
```
